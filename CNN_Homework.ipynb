Process the obtained array from the image with convolution operation. **(20 Points)**

2. You may consider the following steps for this implementation.\
    2.1 Extract all centriods of processing region for each convolution operation.\
    2.2 According to each centroid (i.e., the positions of kernel's center for all convolution operation across input), locate all indices of the elements within the local region for each convolution operation.\
    2.3 Given obtained indices, locate pixel values (i.e., our obtained array elements) and conduct element-wise product between pixel and kernel values.\
    2.4 Sum element-wise product results and assign the value to convolution result at corresponding location.\
    **Note: we did not conduct padding for processed array, and thus, convolution result will become smaller than original array. You may think about the reason.**

```python
import matplotlib.pyplot as plt

######## Convolution with Laplacian Filter ##################
def convolution(image, filter, stride):
  row = image.shape[0]
  col = image[0].size

  # no padding
  outputx = int(((row - filter.shape[0]) / stride) + 1)
  outputy = int(((col - filter.shape[0]) / stride) + 1)
  # print(f"{row ,  col}")
  # print(f"{outputx ,  outputy}")
  output = np.zeros((outputx, outputy))

  for y in range(outputy):
    for x in range(outputx):
      # instead of centroids, calculating a "sliver" of the original image
      xstart = y * stride
      xend = xstart + filter[0].size
      ystart = x * stride
      yend = ystart + filter[0].size

      sliver = image[xstart:xend, ystart:yend]

      output[y, x] = np.sum(sliver * filter)

  return output

def display(image):
  print("First 10 rows:")
  print(image[:10])
  print("Validation for the first 5 x 5 Array:")
  print(image[0:5,0:5])
  print("Output Shape:")
  print(image.shape)

  plt.imshow(image)
  return

laplacian = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
img_matrix = np.array(img_matrix)

filtered_image = convolution(img_matrix, laplacian, 1)
display(filtered_image)
```

```python
######## MaxPooling with the setting of 2X2 ##################
def maxpooling(feature_map):
  row = feature_map.shape[0]
  col = feature_map[0].size

  # note: mask_size = stride
  mask_size = stride = 2

  # https://www.geeksforgeeks.org/deep-learning/cnn-introduction-to-pooling-layer/
  output_y = (row - mask_size) // mask_size + 1
  output_x = (col - mask_size) // mask_size + 1
  output = np.zeros((output_y, output_x))

  for y in range(0, col, stride):
    for x in range(0, row, stride):
      # instead of centroids, calculating a quadrant of the original image
      quadrant = feature_map[y: y + mask_size, x: x + mask_size]

      output[y // stride, x // stride] = np.max(quadrant)
  return output

max_pooing = maxpooling(img_matrix)
display(max_pooing)
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

# Define transformation for training and testing data.
# For training, use RandomCrop, RandomRotation and Normalization with mean 0.1307 variance0.3081
# For testing, just use Normalization.

train_transform = transforms.Compose(
    [torchvision.transforms.RandomCrop(28, 4),
     torchvision.transforms.RandomRotation(15),
     transforms.ToTensor(),
     torchvision.transforms.Normalize(0.1307, 0.3081)])
test_transform = transforms.Compose(
    [transforms.ToTensor(),
      torchvision.transforms.Normalize(0.1307, 0.3081)])

trainset = torchvision.datasets.FashionMNIST("/content/sample_data", train=True,
                                              transform=train_transform, download=True)
testset = torchvision.datasets.FashionMNIST("/content/sample_data", train=False,
                                              transform=test_transform, download=True)

# Load the training and testing datasets from Pytorch
train_data = torch.utils.data.DataLoader(trainset, 128)

test_data = torch.utils.data.DataLoader(testset, 128)


# Print out first image and its correponding label index using matplotlib.
# using the given site as a base
def imshow(img):
  npimg = img.numpy()
  plt.imshow(np.transpose(npimg, (1, 2, 0)))
  plt.show()

dataitr = iter(train_data)
images, labels = next(dataitr)

# show images
imshow(torchvision.utils.make_grid(images[:10]))
# print labels
print(' '.join(f'{labels[i]}' for i in range(10)))

print(f'Number of Training Samples: {len(train_data)}')
print(f'Number of Test Samples: {len(test_data)}')
print(f'Shape of Data: {labels.shape}')
```

### 2.2 Implement LeNet-5 Achitecture

Construct LeNet-5 as learning model for Fashion MNIST classification task. **(15 Points)**

**Tasks:**
1. Build LeNet-5 by creating a new class named 'Net' that inherit nn.Module. Override __init__ and **forward** functions to include necessary network componenets and forward process for neural network.

2. **PRINT OUT** the model summary using torchsummary library (import torchsummary).

**Hints:**
1. You may consider the convolution neural network shown in class as an example. Or https://pytorch.org/tutorials/beginner/blitz/cifar10_tutorial.html.
2. The general structure of LeNet-5 is listed below. Try to fill in the missing parts along with pytorch documentations (https://pytorch.org/docs/stable/index.html).
3. Some architecture settings are listed below.
    - To have 32 X 32 input, you need padding=2 for the first 2D convolution filter (original image is 28 X 28)
    - The kernel size for 2D convolution filter is **5 X 5**. You may think about the reason by calculation.
    - We change tanh activation to **"relu"** activation here.
    - We use MaxPooling instead of original AveragePooling. Please use size 2 X 2 for implementation.
    - Please use **Flatten** to onvert 2D convolution layer to 1D fully connected layer.
    - Gaussian connections are replaced with Softmax, and thus, the outputs are activated by Softmax function based on the number of classes.

4. Validation result:

```python
from torchsummary import summary
# https://www.geeksforgeeks.org/computer-vision/lenet-5-architecture/

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        # convolution layers
        self.conv1 = nn.Conv2d(1, 6, (5, 5), padding=2)
        self.conv2 = nn.Conv2d(6, 16, (5, 5)) # C3

        # flattening layers
        self.fc1 = nn.Linear(16 * 5 * 5, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        # first convolution layer
        x = self.conv1(x)
        x = F.relu(x)
        x = F.max_pool2d(x, (2,2))

        # second convolution layer
        x = self.conv2(x)
        x = F.relu(x)                  # subsampling
        x = F.max_pool2d(x, (2,2))# pooling

        # flattening
        x = torch.flatten(x, 1)
        x = self.fc1(x)
        x = F.relu(x)
        x = self.fc2(x)
        x = F.relu(x)
        output = self.fc3(x)
        return output

# Decide whether you want to run your model on cpu or gpu.
device = 'cuda' if torch.cuda.is_available() else 'cpu'
net = Net().to(device)
summary(net, (1, 28, 28))
```

```python
### Train with SGD optimizer with learning rate =0.1, regularizer=4e-5, momentum=0.9
# https://www.digitalocean.com/community/tutorials/writing-cnns-from-scratch-in-pytorch
# loss function
lossfun = nn.CrossEntropyLoss()

# optimizer
optimizer = torch.optim.SGD(net.parameters(), lr=0.1, momentum=0.9, nesterov=True)

# some other hyperparameters
steps = len(train_data)
epoch = 10

# training
for e in range(epoch):
  total = 0
  correct = 0
  # iterate through all the datapoints in the dataset
  for i, (images, labels) in enumerate(train_data):
    # convert images and labels to specified device
    images = images.to(device)
    labels = labels.to(device)

    # forward pass
    outputs = net(images)
    loss = lossfun(outputs, labels)

    # backward pass
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    # calculating accuracy
    _, predicted = torch.max(outputs.data, 1)
    total += labels.size(0)
    correct += (predicted == labels).sum().item()
  print(f'=== Epoch {e} ===')
  print(f'Training Loss: {loss}')
  print(f'Training Accuracy: {correct / total * 100}')
```

```python
# https://www.geeksforgeeks.org/machine-learning/ml-t-distributed-stochastic-neighbor-embedding-t-sne-algorithm/

net.eval()

all_outputs = []
all_labels = []

with torch.no_grad(): # disable gradient
    for images_batch, labels_batch in test_data:
        images_batch = images_batch.to(device)

        # Get the model's output (logits) for the batch
        outputs_batch = net(images_batch)

        # convert convert tensor to np arrays
        all_outputs.append(outputs_batch.cpu().detach().numpy())
        all_labels.append(labels_batch.cpu().detach().numpy())

# Concatenate all outputs and labels
all_outputs = np.concatenate(all_outputs)
all_labels = np.concatenate(all_labels)

# Apply t-SNE to the collected outputs (logits)
# n_samples = 10000, default perplexity=30 is fine
model = sk.manifold.TSNE(n_components=2, random_state=0, perplexity=30)
tsne_data = model.fit_transform(all_outputs)

tsne_df = pd.DataFrame(data=tsne_data, columns=("Dim_1", "Dim_2"))
tsne_df['label'] = all_labels # add all the labels to the df

plt.scatter(tsne_df['Dim_1'], tsne_df['Dim_2'], c=tsne_df['label'], cmap='tab10', s=10)
plt.title('t-SNE visualization of LeNet-5')
plt.xlabel('Dim 1')
plt.ylabel('Dime2')
plt.show()
```
